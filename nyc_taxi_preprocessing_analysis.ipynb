<!-- SUBSECTION: 5.3 Haversine Distance -->
<div style="background-color: #0f172a;
            border: 1px solid #1e293b;
            border-left: 4px solid #38bdf8;
            border-radius: 10px;
            padding: 20px 22px;

               font-style: italic;">
        5.3 Haversine Distance
    </h3>
    <p style="font-size: 14px; 
              line-height: 1.6; 
              color: #cbd5e1; 
              margin-top: 0; 
              margin-bottom: 0;
              font-style: italic;">
    </p>
</div>

```python
# Vectorized Haversine distance function
def haversine_np(lon1, lat1, lon2, lat2):
    lon1, lat1, lon2, lat2 = map(np.radians, [lon1, lat1, lon2, lat2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = np.sin(dlat / 2.0)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2.0)**2
    c = 2 * np.arcsin(np.sqrt(a))
    km = 6367 * c
    return km

# Calculate distance in kilometers
df['distance_km'] = haversine_np(
    df['pickup_longitude'], df['pickup_latitude'],
    df['dropoff_longitude'], df['dropoff_latitude']
).round(2)

# Display sample results
df[['pickup_longitude', 'pickup_latitude', 'dropoff_longitude', 'dropoff_latitude', 'distance_km']].head()
```

```python
# 1. Calculate Average Speed (km/h)
# Using trip_duration_hrs to compute speed, handling zero duration safely
df['speed_kmh'] = (df['distance_km'] / df['trip_duration_hrs']).round(2)

# Cap extreme or invalid speeds (reasonable city speed threshold: 0 to 120 km/h)
df['speed_kmh'] = df['speed_kmh'].clip(lower=0, upper=120)

# 2. Categorize Distance (Short, Medium, Long)
df['distance_category'] = pd.cut(
    df['distance_km'],
    bins=[-np.inf, 2.5, 10.0, np.inf],
    labels=['Short Trip', 'Medium Trip', 'Long Trip']
)

# 3. Categorize Passenger Count (Solo, Pair, Group)
df['passenger_category'] = pd.cut(
    df['passenger_count'],
    bins=[-np.inf, 1, 2, np.inf],
    labels=['Solo', 'Couple', 'Group']
)

# Display newly engineered attributes
df[['distance_km', 'trip_duration_hrs', 'speed_kmh', 'distance_category', 'passenger_category']].head()
```

```python
import pandas as pd
import plotly.express as px
from IPython.display import HTML

# 1. Re-create simple categorical columns safely
df['passenger_group'] = df['passenger_count'].apply(
    lambda x: 'Solo' if x == 1 else ('Couple' if x == 2 else 'Group')
)

df['time_of_day'] = pd.cut(
    df['pickup_hour'],
    bins=[-1, 6, 12, 18, 24],
    labels=['Night', 'Morning', 'Afternoon', 'Evening']
).astype(str)

df['trip_dist_group'] = pd.cut(
    df['distance_km'],
    bins=[-float('inf'), 2.5, 10.0, float('inf')],
    labels=['Short (<2.5km)', 'Medium (2.5-10km)', 'Long (>10km)']
).astype(str)

# 2. Aggregate counts hierarchically (keeps it super lightweight and fast)
sunburst_df = df.groupby(
    ['passenger_group', 'time_of_day', 'trip_dist_group']
).size().reset_index(name='trip_count')

# 3. Create the Sunburst chart
fig_sunburst = px.sunburst(
    sunburst_df,
    path=['passenger_group', 'time_of_day', 'trip_dist_group'],
    values='trip_count',
    title='NYC Taxi Hierarchical Breakdown (Passenger -> Time -> Distance)'
)

# 4. Display chart directly
HTML(fig_sunburst.to_html(include_plotlyjs='cdn'))
```
